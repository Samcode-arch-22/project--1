```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.25,random_state = 42)
```

```python
x_train.shape
```

```python
### standardization  (
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)#only train data 
x_train
```

```python
##x_test = scaler.fit_transform(x_test) this is wrong couse "fit transform" create a separete mean and standard deviation which create data leakage 
## so test data have to use the mean and standard deviation of train data ( common sense) so here 
x_test = scaler.transform(x_test)
x_test
```

```python
## aplly linear regration 
from sklearn.linear_model import LinearRegression
```

```python
Regression= LinearRegression(n_jobs=1)
```

```python
Regression.fit(x_train,y_train)
```

```python
## plot best fit line in terms of traning data 
plt.scatter(x_train,y_train)
plt.plot(x_train,Regression.predict(x_train))
```

```python
plt.plot(x_train,Regression.predict(x_train))
```

```python
## prediction for test data 
y_pred = Regression.predict(x_test)
y_pred
```
